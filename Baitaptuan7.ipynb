Bài 1 : Viết hàm tìm kiếm nhị phân
trên một danh sách đã
được sắp xếp tăng dần.
Trả về chỉ số của phần tử
hoặc -1 nếu không tìm
thấy. mẫu hàm: def
binary_search(arr, left,
right, target):



```python
arr = list(map(int,input().split()))
target= int(input())
left= 0
right= len(arr)
def binary_search(arr, left, right, target):
    for i in arr:
        if i == target:
            break
    else:
        return -1
    
    mid= (left+right)//2
    if arr[mid] == target:
        return mid
    elif arr[mid] > target:
        return binary_search(arr, left, mid-1, target)
    elif arr[mid] < target:
        return binary_search(arr, mid+1, right, target)
print(binary_search(arr, left, right, target))
```

Bài 2: Cho danh sách arr và một
số nguyên x, đếm xem x
xuất hiện bao nhiêu lần
trong danh sách. def
count_occurrences(arr, x) -
> int:

```python
arr=list(map(int,input().split()))
x= int(input())
def count_occurrences(arr,x):
    dem=0
    for i in arr:
        if i==x:
            dem+=1
    return dem
print(count_occurrences(arr,x))
```

Baif 3: Viết hàm thực hiện thuật
toán bubble sort, Selection
Sort để sắp xếp danh sách
theo thứ tự tăng dần. Sắp
xếp nổi bọt (Bubble Sort)

```python
arr= list(map(int, input().split()))
def bubble_sort(arr):
    for i in range(len(arr)):
        for j in range(len(arr)-i -1):
            if arr[j] > arr[j+1]:
                arr[j],arr[j+1]= arr[j+1], arr[j]
    return arr


def selection_sort(arr):
    delta=[]
    for i in range(len(arr)):
        delta.append(min(arr))
        arr.remove(min(arr))
    return delta
print(*selection_sort(arr))
```
