Bài 1: Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu
mỗi phần tử x trong mảng xuất hiện đúng x lần.
Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa
đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp.



```python
from collections import Counter # Counter đếm số phần tử trong list và tạo dict 

nums= list(map(int, input().split()))
cnt=(Counter(nums))
remove=0
for k,v in cnt.items():
    if v>k:
        remove+= v-k
    elif k>v:
        remove +=v
if remove ==0:
    print('mang dep')
else:
    print(f"neu bo di {remove} phan tu thi nums la mang dep")
 
```

Bài 2: Tìm phần tử xuất hiện nhiều hơn n/2 lần trong mảng (nếu có) với độ
phức tạp thời gian là O(n), độ phức tạp bộ nhớ là O(1), - Boyer–Moore.
Nếu không có phần tử nào thỏa mãn thì không in gì


```python
from collections import Counter 
del list
nums= list(map(int,input().split()))
cnt= Counter(nums)
n= len(nums)//2
list=[]
for k,v in cnt.items():
    if v> n:
        list.append(k)
if list == []:
    print('None')
else:
    print(*list)
```

Bài 3: Sắp xếp có điều kiện: Số chẵn đứng trước, số lẻ đứng sau, chẵn tăng
dần, lẻ giảm dần; 

```python
nums= list(map(int, input().split()))
delta_chan=[]
delta_le=[]
for i in nums:
    if i%2==0:
        delta_chan.append(i)
    else:
        delta_le.append(i)
a= sorted(delta_chan)
b= sorted(delta_le, reverse= True)
c= a+b
print(*c)


```

Bài 7: Cho N tên sinh viên gồm cả họ và tên, tìm xem tên nào hay gặp nhất.
Nếu có nhiều tên cùng số lần gặp thì lấy tên có thứ tự từ điển nhỏ nhất

```python
from collections import Counter
n= int(input())
new_list=[]
for i in range(n):
    ho_va_ten= input().split()
    name= ho_va_ten[-1]
    new_list.append(name)

max_name=[]
cnt= Counter(new_list)
for k,v in cnt.items():
    if v == max(cnt.values()):
        max_name.append(k)
print(min(max_name))


```

Bài 12:Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra xem dãy ngoặc đã cho có hợp lệ
hay không

```python
s= input()
stack =[]
for i in s:
    if i in "({[":
        stack.append(i)
    else:
        if len(stack) == 0:
            print('Fail')
            break

        top= stack.pop()
        if top == "(" and i != ")":
            print('Fail')
            break
        elif top== "{" and i != "}":
            print('Fail')
            break
        elif top == "[" and i != "]":
            print('Fail')
            break

else:
    if len(stack) == 0:
        print('Success')
    else:
        print('Fail')
```
