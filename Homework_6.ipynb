Bài 2: Nhập vào mảng vuông 2 chiều n hàng, n cột, in ra đường chéo chính và
đường chéo phụ của mảng


```python
# Nhập ma trận
n= int(input())
a=[]
for i in range(n):
    row= list(map(int,input().split()))
    a.append(row)

# Đường chéo chính khi i==j
for i in range(n):
    print(a[i][i], end="")

print()

# Đùng chéo phụ i+j == n-1
for i in range(n):
    print(a[i][n-i-1], end="")
```

Bài 3: Viết hàm đưa vào 1 list số nguyên, tìm và trả về vị trí có giá trị lớn nhất
trong list

```python
lst= list(map(int, input().split()))
def maximize(lst):
    k=0
    for i in lst:
        if i>k:
            k=i
    return k
print(maximize(lst))

```

Bài 4: Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và
trả về vị trí của phần tử đầu tiên có giá trị k trong list, nếu không có thì trả về
-1


```python
lst= list(map(int, input().split()))
k= int(input())
def locate(lst,k):
    dem=0
    for i in lst:
        dem +=1
        if i ==k:
            break
    return dem
print(locate(lst,k))
```

Bài 6: Viết hàm đưa vào 1 list số nguyên, tìm và trả về vị trí có giá trị lớn nhất
trong list

```python
lst= list(map(int,input().split()))
def locate_max(lst):
    dem=0
    for i in lst:
        dem+=1
        if i == max(lst):
            break
    return dem
print(locate_max(lst))
```

Bài 7: VIết hàm xóa đi các số <x có trong mảng

```python
lst= list(map(int,input().split()))
x= int(input())
new_list=[]
def delete_number(lst,x):
    for i in lst:
        if i>x:
            new_list.append(i)
    return new_list
print(delete_number(lst,x))
```

Bài 8: Viêết hàm đẩy tất cả các số 0 về đầu (về cuối), các số khác vẫn đúng vị trí
trước sau: 0 2 3 2 4 0 0 0 9 => 0 0 0 2 3 2 4 9

```python
x= int(input())
def swap_zeros(x):
    new_number=''
    for i in str(x):
        if int(i) !=0:
            new_number+= str(i)
    while not len(str(x)) == len(new_number):
        new_number= '0'+new_number
    return new_number
print(swap_zeros(x))
```
